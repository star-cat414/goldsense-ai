# GoldSense AI — Model Comparison

Experimental workspace for comparing XGBoost vs Linear Regression. The official implementation is
`backend/app/ml/train.py` with evaluation in `backend/app/ml/evaluate.py`.

## Official methodology
- **Expanding-window walk-forward, one-day-ahead** on the chronological test window
  (2025-02-19 → 2026-09-10, 387 points)
- Both models scored on the **identical** window; results in
  `backend/saved_models/evaluation.json`
- Current results:
  | Model | MAE | RMSE | MAPE | R² |
  |---|---|---|---|---|
  | XGBoost | 76.47 | 104.26 | 1.86% | 0.9737 |
  | Linear Regression | 51.21 | 76.06 | 1.24% | 0.986 |
