# GoldSense AI — Feature Engineering

Prototyping notebook for the 35-feature pipeline used in production. The reference implementation
lives in `backend/app/ml/preprocess.py` (`engineer_features`, `split_chronological`).

## Features built from `gold_daily.csv`
- **Lags** — `close_lag_{1,2,3,7,14,30}`, volume lags
- **Moving averages** — MA 7/14/30/50 and price-vs-MA ratios
- **Volatility** — rolling std 7/14, ATR-style measures
- **Momentum** — daily change %, RSI-14, MACD line/signal/histogram, EMA 12/26
- **Bollinger Bands** — upper/lower, width, %B
- **Market indicators** — silver/oil/S&P500/USD-index close and pct-change lags
- **Calendar** — day of week / month cyclical encoding
